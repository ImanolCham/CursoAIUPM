## Descripción de los datos

El fichero de datos contiene la información de 303 pacientes con 14 variables que son características asociadas con enfermedad cardíaca. Son 

* **age:** Edad del paciente
* **sex:** Sexo biológico del paciente (0 = Hombre; 1 = Mujer)
* **cp:** Tipo de dolor de pecho (0: Dolor característico de angina de pecho (pecho, brazo, mandíbula); 1: Dolor atípico de angina de pecho (espalda, epigástrico); 2: Dolor no asociado a angina de pecho; 3:Asintomático )
*  **trestbps** Presión sanguínea (alta) en reposo en mmHg.
*  **chol** Colesterol sérico total en sangre (HDL + LDL + triglicéridos) en mg/dl
*  **fbs** Nivel de glucosa en sangre (0: normal (<120mg/dl); 1: alto (>1120mg/dl)).
*  **restecg** Resultado del electrocardiograma en reposo (0: Sin hallazgos significativos; 1: Resultados anómalos  de onda ST-T (inversiones de onda T y / o elevación o Depresión de ST de > 005 mV); 2: Signos de hipertrofia del ventrículo izquierdo según criterios de Estes)
*  **thalach** Frecuencia cardíaca máxima en esfuerzo en lpm.
*  **exang** Angina de pecho provocada por ejercicio. (0: False; 1: True)
*  **oldpeak** Depresión en el segmento ST inducida por el ejercicio respecto a la situación de reposo.
*  **slope** Tipo de pendiente de la onda ST (0: Incremento positivo; 1:Sin pendiente; 2: Incremento negativo)
  
* **ca** Número de vasos principales coloreados por la fluoroscopía (0-4).
* **thal** Estado del corazón segun la prueba de Thallium (0: Normal; 1: Daño permanente; Daño reversible; 3: Daño no especificado).
* **target** Variable objetivo a predecir (0: Sin enfermedad cardíaca; 1. Enfermedad cardíaca)
