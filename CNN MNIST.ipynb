# CNN para clasificar MNIST

Cargamos librerias y el dataset.

Normalizamos las imágenes, redimensionamos a un canal y transformamos las etiquetas

```python
import keras
from keras import layers, models
from keras.datasets import mnist
from keras.utils import to_categorical

# 1. Cargar y preprocesar el dataset MNIST
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Normalizar los valores de píxeles al rango [0, 1]
x_train = x_train.astype('float32') / 255
x_test = x_test.astype('float32') / 255

# Redimensionar las imágenes para que tengan un canal (escala de grises). Si fuera RGB, sería (28, 28, 3).
x_train = x_train.reshape((x_train.shape[0], 28, 28, 1))
x_test = x_test.reshape((x_test.shape[0], 28, 28, 1))

# Convertir las etiquetas a one-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)
```

```python
print('x_train shape:', x_train.shape)
print('x_test shape:', x_test.shape)    
print('y_train shape:', y_train.shape)
print('y_test shape:', y_test.shape)
```

```python
# 2. Definir la arquitectura del modelo CNN
model = models.Sequential()
# Respecto al tamaño, el 3,3 es el primer filtro, que mira 3 pixeles por 3 pixeles, y el 32 es el número de filtros que se aplican a la imagen. La función de activación ReLU introduce no linealidad al modelo, lo que permite aprender patrones más complejos. La forma de entrada (28, 28, 1) indica que las imágenes son de 28x28 píxeles y tienen un solo canal (escala de grises).
model.add(layers.Conv2D(32, (3, 3), activation='relu', input_shape=(28, 28, 1))) # aquí se le añade la red convolucional con 32 filtros de tamaño 3x3, función de activación ReLU y la forma de entrada (28, 28, 1)
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Conv2D(64, (3, 3), activation='relu'))
model.add(layers.MaxPooling2D((2, 2)))
model.add(layers.Flatten())
model.add(layers.Dense(10, activation='softmax'))

# 3. Compilar el modelo
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 4. Entrenar el modelo
model.fit(x_train, y_train, epochs=5, batch_size=64, validation_data=(x_test, y_test))

# 5. Evaluar el modelo
loss, accuracy = model.evaluate(x_test, y_test, verbose=0)
print('Test loss:', loss)
print('Test accuracy:', accuracy)
```
