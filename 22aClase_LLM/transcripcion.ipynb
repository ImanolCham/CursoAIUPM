# Whisper

```python
import openai
import os
from openai import OpenAI
import textwrap

openai.api_key = os.getenv("OPENAI_API_KEY")
client = OpenAI()
```

```python
# Ruta del archivo de audio a transcribir
audio_file_path = "audioATC.wav"  # Reemplaza con la ruta a tu archivo de audio
audio_file= open(audio_file_path, "rb")

transcription = client.audio.transcriptions.create(
    model="gpt-4o-transcribe", 
    file=audio_file,
    language="es",
    response_format="text",
    
    
)

# Extraer el texto transcrito
formatted_text = textwrap.fill(transcription, width=80)
print("Transcripción:")
print(formatted_text)



# Crear un prompt para solicitar un resumen de la transcripción
prompt = f"Por favor, resume, en inglés, el siguiente texto en 30 palabras:\n\n{transcription}"

# Utilizar el modelo para generar el resumen

response = client.responses.create(
    model="gpt-4o-mini",
    input=prompt
)

# Extraer y mostrar el resumen

print("\nResumen:")

formatted_text = textwrap.fill(response.output_text, width=80)
print(formatted_text)
```
