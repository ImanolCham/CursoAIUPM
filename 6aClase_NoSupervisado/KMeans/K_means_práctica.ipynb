Para estudiar la bondad del método vamos a emplear 3 métricas distintas:

* Inercia-WCSS (`kmeans.inertia_`)
    - Representa la suma de las distancias cuadráticas entre cada punto y el centroide de su clúster.  
    - Cuanto menor sea la inercia, mejor se ajustan los puntos a sus respectivos centroides.  
    - Se utiliza, por ejemplo, para construir la curva del codo, que ayuda a elegir el número óptimo de clústeres `k`.


* ARI — Adjusted Rand Index
    - Mide la similitud entre las etiquetas reales (`y_true`) y las obtenidas por el modelo (`cluster_labels`).  
    - Su valor está ajustado para que un agrupamiento aleatorio dé un resultado cercano a 0.

| Valor de ARI | Interpretación |
|:--------------:|:---------------|
| **1.0** | Agrupamiento idéntico a las clases reales |
| **0.0** | Agrupamiento sin relación (aleatorio) |
| **< 0.0** | Peor que aleatorio |


* Coeficiente de Silhouette

    - Evalúa la cohesión interna y la separación entre clústeres sin usar etiquetas reales.  
    - Sus valores posibles están en el rango $[-1, 1]$:
        - Cerca de 1: los puntos están bien agrupados y separados entre sí.  
        - Cerca de 0: los puntos están en la frontera entre clústeres.  
        - Negativo: posibles asignaciones incorrectas.


Estas tres métricas combinadas permiten analizar la calidad del agrupamiento:
- **Inercia baja** → clústeres compactos.  
- **ARI alto** → buena correspondencia con las clases reales (solo si se conocen).  
- **Silhouette alto** → clústeres bien definidos y separados.
